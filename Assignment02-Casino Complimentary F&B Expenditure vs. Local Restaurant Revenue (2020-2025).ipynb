```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('Macau_Casino_Food_2020_2025.xlsx')
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))
color = 'tab:red'
ax1.set_xlabel('Year')
ax1.set_ylabel('Casino F&B Exp (Million MOP)', color=color)
ax1.plot(df['Year'], df['Casino_FoodBev_Exp (Million)'], color=color, marker='o', linewidth=2, label='Casino F&B Exp')
ax1.tick_params(axis='y', labelcolor=color)
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Local F&B Revenue (Million MOP)', color=color)
ax2.plot(df['Year'], df['Local_Revenue (Million)'], color=color, marker='s', linewidth=2, label='Local F&B Revenue')
ax2.tick_params(axis='y', labelcolor=color)
plt.title('Casino Complimentary F&B Expenditure vs. Local Restaurant Revenue (2020-2025)')
fig.tight_layout()
plt.grid(True)
plt.axvline(x=2023, color='gray', linestyle='--')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('Macau_Casino_Food_2020_2025.xlsx')

fig, (ax1, ax3) = plt.subplots(2, 1, figsize=(12, 8), gridspec_kw={'height_ratios': [3, 1]}, sharex=True)

color = 'tab:red'
ax1.set_ylabel('Casino F&B Exp (Million MOP)', color=color)
ax1.plot(df['Year'], df['Casino_FoodBev_Exp (Million)'], color=color, marker='o', linewidth=2, label='Casino F&B Exp')
ax1.tick_params(axis='y', labelcolor=color)


ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Local F&B Revenue (Million MOP)', color=color)
ax2.plot(df['Year'], df['Local_Revenue (Million)'], color=color, marker='s', linewidth=2, label='Local F&B Revenue')
ax2.tick_params(axis='y', labelcolor=color)

ax1.axvline(x=2023, color='gray', linestyle='--', alpha=0.7)
ax1.text(2023.1, 7000, 'Post-Pandemic\nRecovery', color='gray', fontsize=10, verticalalignment='top')

ax1.axvline(x=2024, color='orange', linestyle='--', alpha=0.7)
ax1.text(2024.1, 7000, 'Regulatory\nTightening', color='orange', fontsize=10, verticalalignment='top')


ax1.set_title('Casino Complimentary F&B Expenditure vs. Local Restaurant Revenue (2020-2025)')
ax1.grid(True)


colors = ['red' if x < 0 else 'green' for x in df['Local_Profit (Million)']]
ax3.bar(df['Year'], df['Local_Profit (Million)'], color=colors, width=0.5)


ax3.axhline(0, color='black', linewidth=0.8, linestyle='-')

ax3.set_ylabel('Local Profit\n(Million MOP)')
ax3.set_xlabel('Year')


fig.tight_layout()
plt.show()
```
